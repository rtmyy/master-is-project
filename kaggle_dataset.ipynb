### 🧱 การจัดเตรียมข้อมูลและการแบ่งข้อมูลเป็น Train และ Test set
ก่อนอื่นเราจะกำหนดตัวแปรต้น (Features) และตัวแปรเป้าหมาย (Target) รวมถึงแปลงตัวแปรเป้าหมาย `Attrition` ให้เป็นตัวเลข (Yes = 1, No = 0) จากนั้นใช้ฟังก์ชัน `train_test_split` โดยกำหนดพารามิเตอร์ `stratify=y` เพื่อให้มั่นใจว่าสัดส่วนของกลุ่มคนลาออก (Class Distribution) ในข้อมูล Train และ Test ยังคงเท่ากัน

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer
import pandas as pd
import numpy as np

# 1. แปลงตัวแปรเป้าหมาย (Target)
y = df['Attrition'].apply(lambda x: 1 if x == 'Yes' else 0)

# 2. คัดเลือก Features และลบคอลัมน์ที่ไม่ส่งผลเชิงทำนายออก
columns_to_drop = ['Attrition', 'EmployeeNumber', 'EmployeeCount', 'Over18', 'StandardHours']
X_raw = df.drop(columns=[col for col in columns_to_drop if col in df.columns])

# 3. แบ่งชุดข้อมูลดิบเป็น Train และ Test (stratify ป้องกันความไม่สมดุล)
X_train_raw, X_test_raw, y_train, y_test = train_test_split(
    X_raw, y, test_size=0.2, random_state=42, stratify=y
)

# 4. ระบุคอลัมน์ประเภทหมวดหมู่ (Categorical Features) เพื่อทำ One-Hot Encoding
categorical_cols = X_train_raw.select_dtypes(include=['object', 'category']).columns.tolist()
numeric_cols = X_train_raw.select_dtypes(exclude=['object', 'category']).columns.tolist()

# 5. ใช้ ColumnTransformer เพื่อแปลง OneHotEncoder โดยฝึกสอน (fit) เฉพาะใน Train Set เท่านั้น
preprocessor = ColumnTransformer(
    transformers=[
        ('cat', OneHotEncoder(drop='first', sparse_output=False, handle_unknown='ignore'), categorical_cols)
    ],
    remainder='passthrough'
)

# ฟิตข้อมูลและแปลงค่าเพื่อสร้าง Features สำหรับเทรนและทดสอบ
X_train_arr = preprocessor.fit_transform(X_train_raw)
X_test_arr = preprocessor.transform(X_test_raw)

# ดึงชื่อฟีเจอร์หลังจากการทำแปลงข้อมูล One-Hot
encoded_cat_features = preprocessor.named_transformers_['cat'].get_feature_names_out(categorical_cols).tolist()
all_feature_names = encoded_cat_features + numeric_cols

# แปลงผลลัพธ์กลับเป็น pandas DataFrame เพื่อความสะดวกในการใช้งานกับ SHAP และวิเคราะห์ต่อ
X_train = pd.DataFrame(X_train_arr, columns=all_feature_names, index=X_train_raw.index)
X_test = pd.DataFrame(X_test_arr, columns=all_feature_names, index=X_test_raw.index)

print(f"จำนวนคอลัมน์หลังทำแปลงแบบไร้ Data Leakage: {X_train.shape[1]}")
print(f"พารามิเตอร์ OneHot ถูกเรียนรู้เฉพาะใน Train set สำเร็จ!")
```

```python
from imblearn.over_sampling import SMOTE
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report, precision_recall_curve, auc
import pandas as pd

# 1. ทำการ Oversampling ด้วย SMOTE บนข้อมูลที่ผ่าน ColumnTransformer (Train Set เท่านั้น)
smote = SMOTE(random_state=42)
X_train_resampled, y_train_resampled = smote.fit_resample(X_train, y_train)

# 2. ฝึกโมเดล Random Forest ด้วยข้อมูลที่ป้องกัน Data Leakage และปรับสมดุลแล้ว
smote_rf = RandomForestClassifier(random_state=42)
smote_rf.fit(X_train_resampled, y_train_resampled)

# 3. ทำนายผลบน Test Set
y_pred_smote = smote_rf.predict(X_test)
y_probs_smote = smote_rf.predict_proba(X_test)[:, 1]

# 4. แสดงรายงานผลการประเมิน
print("=== SMOTE + Random Forest (Clean Pipeline) Classification Report ===")
print(classification_report(y_test, y_pred_smote, target_names=['No (0)', 'Yes (1)']))

precision_s, recall_s, _ = precision_recall_curve(y_test, y_probs_smote)
pr_auc_smote = auc(recall_s, precision_s)
print(f"PR-AUC Score (SMOTE - Leakage Free): {pr_auc_smote:.4f}")
```

```python
# 5. คัดเลือกกลุ่มพนักงานที่มีความเสี่ยงสูง (High Risk) ด้วยระดับเกณฑ์เดิม Threshold >= 50%
test_analysis_df = X_test.copy()
test_analysis_df['Actual_Attrition'] = y_test
test_analysis_df['Predicted_Probability'] = y_probs_smote
test_analysis_df['Predicted_Attrition'] = y_pred_smote

high_risk_employees = test_analysis_df[test_analysis_df['Predicted_Probability'] >= 0.5].sort_values(by='Predicted_Probability', ascending=False)
print(f"=== จำนวนพนักงานในกลุ่มเสี่ยงสูงหลังปรับปรุง Preprocessing: {high_risk_employees.shape[0]} คน ===")
```

```python
from sklearn.metrics import classification_report, confusion_matrix, precision_recall_curve, auc
import pandas as pd

# ฟังก์ชันช่วยสรุปผลลัพธ์ของแต่ละโมเดลอย่างละเอียด
def evaluate_model_pipeline(model, name, X_t, y_t):
    preds = model.predict(X_t)
    probs = model.predict_proba(X_t)[:, 1]

    # Confusion Matrix
    cm = confusion_matrix(y_t, preds)
    tn, fp, fn, tp = cm.ravel()

    # Precision-Recall AUC
    precision, recall, _ = precision_recall_curve(y_t, probs)
    pr_auc_val = auc(recall, precision)

    # ดึงค่า Recall และ F1-score ของ Class 1
    report = classification_report(y_t, preds, output_dict=True)
    rec_1 = report['1']['recall']
    f1_1 = report['1']['f1-score']

    print(f"=== {name} ===")
    print(f"Confusion Matrix: TN={tn}, FP={fp}, FN={fn}, TP={tp}")
    print(f"Recall (Class 1): {rec_1:.2f}")
    print(f"F1-Score (Class 1): {f1_1:.2f}")
    print(f"PR-AUC Score: {pr_auc_val:.4f}")
    print("-"*40)
    return tn, fp, fn, tp, rec_1, f1_1, pr_auc_val

# ประเมินผล 3 โมเดลในกรอบวิธีใหม่บนชุดข้อมูลที่ป้องกัน Leakage แล้ว
_ = evaluate_model_pipeline(baseline_model, "1. Baseline RF", X_test, y_test)
_ = evaluate_model_pipeline(cost_sensitive_model, "2. Cost-sensitive RF", X_test, y_test)
_ = evaluate_model_pipeline(partial_under_model, "3. Partial Undersampling RF", X_test, y_test)

# ยืนยันจำนวนพนักงานที่มี Probability >= 0.50 สำหรับ Partial Undersampling RF
prob_under = partial_under_model.predict_proba(X_test)[:, 1]
risk_count = (prob_under >= 0.50).sum()
print(f"จำนวนพนักงานกลุ่มเสี่ยงสูง (Prob >= 50%) จากโมเดล Partial Undersampling RF: {risk_count} คน")
```

### 🧪 การแก้ปัญหาโดยใช้ SMOTE (Synthetic Minority Over-sampling Technique)
เราจะใช้ไลบรารี `imblearn` เพื่อทำ Oversampling เฉพาะในชุดข้อมูลสอน (Train Set) เท่านั้น (เพื่อป้องกันปัญหา Data Leakage) แล้วนำข้อมูลที่สมบูรณ์ไปเทรนโมเดล Random Forest

```python
# ติดตั้ง imbalanced-learn หากยังไม่มีในสภาพแวดล้อม
!pip install -q imbalanced-learn

from imblearn.over_sampling import SMOTE

# 1. ทำการ Oversampling ด้วย SMOTE บนข้อมูล Train Set เท่านั้น
smote = SMOTE(random_state=42)
X_train_resampled, y_train_resampled = smote.fit_resample(X_train, y_train)

print("=== หลังทำ SMOTE ===")
print(f"จำนวนแถวข้อมูลสอนหลังทำ SMOTE: {X_train_resampled.shape[0]} แถว")
print("สัดส่วนคลาสหลังทำ SMOTE:")
print(pd.Series(y_train_resampled).value_counts(normalize=True) * 100)

# 2. ฝึกโมเดล Random Forest ด้วยข้อมูลที่ปรับสมดุลแล้ว
smote_rf = RandomForestClassifier(random_state=42)
smote_rf.fit(X_train_resampled, y_train_resampled)

# 3. ทำนายผล
y_pred_smote = smote_rf.predict(X_test)
y_probs_smote = smote_rf.predict_proba(X_test)[:, 1]

# 4. แสดงรายงานผล
print("\n=== SMOTE + Random Forest Classification Report ===")
print(classification_report(y_test, y_pred_smote, target_names=['No (0)', 'Yes (1)']))

# 5. คำนวณ PR-AUC
precision_s, recall_s, _ = precision_recall_curve(y_test, y_probs_smote)
pr_auc_smote = auc(recall_s, precision_s)
print(f"PR-AUC Score (SMOTE): {pr_auc_smote:.4f}")
```

### 📋 การสร้าง Rule-based Recommendations สำหรับป้องกันการลาออกของพนักงานกลุ่มเสี่ยง
เราจะกำหนดเกณฑ์และกฎ (Rules) อัตโนมัติโดยอิงจากปัจจัยสำคัญที่มักส่งผลต่อการตัดสินใจลาออก (เช่น รายได้ต่ำ, ทำโอเวอร์ไทม์หนัก, หรือต้องเดินทางไกล) เพื่อเสนอแนะแนวทางแก้ไขปัญหาที่เจาะจงรายบุคคลให้กับฝ่ายบุคคล (HR)

```python
def generate_hr_recommendation(row):
    recommendations = []

    # กฎข้อที่ 1: การทำโอเวอร์ไทม์ (OverTime)
    if 'OverTime_Yes' in row and row['OverTime_Yes'] == 1:
        recommendations.append("🔴 ลดภาระงานนอกเวลา (Overtime Redesign): ตรวจสอบการทำโอเวอร์ไทม์ และจัดสรรงานหรือปรับแผนงานเพื่อลดภาระงานหนัก")

    # กฎข้อที่ 2: รายได้ต่อเดือนต่ำกว่าเกณฑ์เฉลี่ยพนักงานทั่วไป
    if row['MonthlyIncome'] < 3000:
        recommendations.append("💵 ทบทวนโครงสร้างรายได้ (Compensation Review): พิจารณาปรับปรุงฐานเงินเดือนขั้นต่ำ หรือเพิ่มเบี้ยเลี้ยงพิเศษ")

    # กฎข้อที่ 3: เดินทางไกล
    if row['DistanceFromHome'] > 15:
        recommendations.append("🚗 นโยบายการทำงานยืดหยุ่น (Flexible Work): เสนอทางเลือกในการทำงานจากบ้าน (Work from Home) หรือความยืดหยุ่นในการเดินทาง")

    # กฎข้อที่ 4: ความพึงพอใจสิ่งแวดล้อมต่ำ
    if row['EnvironmentSatisfaction'] <= 2:
        recommendations.append("🏢 ปรับปรุงความสัมพันธ์และบรรยากาศ (Workplace Environment Improvement): พูดคุยเปิดอกเพื่อแก้ไขปัญหาในทีมหรือสภาพแวดล้อม")

    if len(recommendations) == 0:
        recommendations.append("🟢 พูดคุยและติดตาม (Career Development Check-in): ร่วมพูดคุยเพื่อวางแผนความก้าวหน้าในอาชีพ")

    return "\n".join([f"{i+1}. {rec}" for i, rec in enumerate(recommendations)])

# นำกฎที่สร้างขึ้นมาใช้กับพนักงานกลุ่มเสี่ยงสูง (High Risk)
risk_report = high_risk_employees.copy()
risk_report['Actionable_Recommendations'] = risk_report.apply(generate_hr_recommendation, axis=1)

# แสดงรายงานแนะนำการแก้ไขปัญหาของพนักงานที่มีความเสี่ยงสูงที่สุด 3 อันดับแรก
print("=== รายงานคำแนะนำในการดูแลรักษาพนักงานกลุ่มเสี่ยงสูงสุด (Top 3 Candidates) ===")
for idx, row in risk_report.head(3).iterrows():
    print(f"\nรหัสพนักงาน (Index): {idx}")
    print(f"- ความเสี่ยงในการลาออก: {row['Predicted_Probability']*100:.1f}%")
    print(f"- อายุ: {row['Age']} ปี | รายได้ต่อเดือน: ${row['MonthlyIncome']:.0f} | ระยะทางบ้าน: {row['DistanceFromHome']} ไมล์")
    print(f"- คำแนะนำสำหรับฝ่าย HR:\n{row['Actionable_Recommendations']}")
    print("-" * 80)
```

### 🧪 การค้นหา Decision Threshold ที่เหมาะสมที่สุดด้วยสมมติฐาน Cost Ratio และการทำ Cross-Validation
ในขั้นตอนนี้ เราจะทำการค้นหาจุดตัดสินใจ (Threshold) ที่ให้ **Expected Cost ต่ำที่สุด** บนชุดข้อมูลตรวจสอบ (Validation Folds) โดยมีวิธีการคำนวณ Expected Cost ดังนี้:
$$\text{Expected Cost} = (\text{Cost}_{FN} \times FN) + (\text{Cost}_{FP} \times FP)$$

โดยกำหนดสมมติฐานของอัตราส่วนต้นทุน (FN:FP) ไว้ 4 รูปแบบคือ **1:1, 2:1, 5:1, และ 10:1**

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import StratifiedKFold
from sklearn.metrics import confusion_matrix, precision_recall_curve, classification_report, average_precision_score
from imblearn.pipeline import Pipeline as ImbPipeline
from imblearn.over_sampling import SMOTE
from sklearn.ensemble import RandomForestClassifier

# 1. กำหนดรูปแบบอัตราส่วนต้นทุน (Cost Ratios)
cost_scenarios = [
    {"ratio": "1:1", "cost_fn": 1, "cost_fp": 1},
    {"ratio": "2:1", "cost_fn": 2, "cost_fp": 1},
    {"ratio": "5:1", "cost_fn": 5, "cost_fp": 1},
    {"ratio": "10:1", "cost_fn": 10, "cost_fp": 1}
]

# 2. ทำ Stratified 5-Fold Cross-Validation บน Train Set เพื่อเลือก Threshold
skf = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
threshold_grid = np.linspace(0.01, 0.99, 99)

# พจนานุกรมสำหรับบันทึก Expected Cost ของแต่ละ Threshold ในแต่ละคู่ Cost Scenario
cv_cost_results = {scen["ratio"]: [] for scen in cost_scenarios}

# สร้าง Pipeline ที่มีการประยุกต์ใช้ SMOTE เฉพาะในฝั่ง Train Fold เท่านั้น (ป้องกัน Leakage)
for train_idx, val_idx in skf.split(X_train, y_train):
    X_tr_fold, X_val_fold = X_train.iloc[train_idx], X_train.iloc[val_idx]
    y_tr_fold, y_val_fold = y_train.iloc[train_idx], y_train.iloc[val_idx]

    # สร้าง Pipeline
    pipeline = ImbPipeline([
        ('smote', SMOTE(random_state=42)),
        ('rf', RandomForestClassifier(random_state=42))
    ])

    # เทรนโมเดล
    pipeline.fit(X_tr_fold, y_tr_fold)

    # ทำนายความน่าจะเป็นของ Validation Fold
    val_probs = pipeline.predict_proba(X_val_fold)[:, 1]

    # ประเมินค่าใช้จ่ายที่เกิดจาก Threshold แต่ละระดับ
    for scen in cost_scenarios:
        ratio_name = scen["ratio"]
        cfn = scen["cost_fn"]
        cfp = scen["cost_fp"]

        fold_costs = []
        for th in threshold_grid:
            preds = (val_probs >= th).astype(int)
            tn, fp, fn, tp = confusion_matrix(y_val_fold, preds).ravel()
            expected_cost = (cfn * fn) + (cfp * fp)
            fold_costs.append(expected_cost)

        cv_cost_results[ratio_name].append(fold_costs)

# 3. สรุปผลเพื่อหา Optimal Threshold สำหรับแต่ละ Cost Scenario
optimal_thresholds = {}
print("=== [สรุปผลการวิเคราะห์บน Validation (CV)] ===")
for scen in cost_scenarios:
    ratio_name = scen["ratio"]
    costs_array = np.array(cv_cost_results[ratio_name]) # มิติ: (5 folds, 99 thresholds)

    # คำนวณค่าเฉลี่ยและส่วนเบี่ยงเบนมาตรฐานของค่าใช้จ่ายในแต่ละระดับ Threshold
    mean_costs = np.mean(costs_array, axis=0)
    std_costs = np.std(costs_array, axis=0)

    best_idx = np.argmin(mean_costs)
    best_th = threshold_grid[best_idx]
    best_cost = mean_costs[best_idx]
    best_cost_std = std_costs[best_idx]

    optimal_thresholds[ratio_name] = best_th
    print(f"Cost Ratio {ratio_name} -> Optimal Threshold: {best_th:.2f} | Expected Cost เฉลี่ย: {best_cost:.2f} (± {best_cost_std:.2f})")
```

### 🎯 การตรึงจุดตัดสินใจ (Freeze Optimal Thresholds) และประเมินผลแบบนัดเดียวบน Holdout Test Set
ในขั้นตอนนี้ เราจะทำการฝีกสอนโมเดลปลายทางบนชุดข้อมูลสอนทั้งหมด (`X_train`) และนำมาประเมินประสิทธิภาพบนชุดข้อมูลทดสอบ (`X_test`) ด้วยเกณฑ์ตัดสินใจ (Thresholds) ที่ถูกเลือกมาจาก Cross-Validation เพื่อป้องกันการเกิดปัญหา Optimistic Bias ในการประเมินผล

```python
# 1. เทรนโมเดลตัวเต็มบนชุดข้อมูลสอน
final_pipeline = ImbPipeline([
    ('smote', SMOTE(random_state=42)),
    ('rf', RandomForestClassifier(random_state=42))
])
final_pipeline.fit(X_train, y_train)

# 2. ทำนายความน่าจะเป็นของชุดข้อมูลทดสอบ (Holdout Test Set)
test_probs = final_pipeline.predict_proba(X_test)[:, 1]

# 3. คำนวณค่าคงที่ Average Precision (AP) ของชุดทดสอบ
ap_score_test = average_precision_score(y_test, test_probs)

# 4. ประเมินผลลัพธ์ของแต่ละ Threshold บน Holdout Test Set
test_evaluation_results = []

# เริ่มต้นเพิ่ม Baseline Threshold 0.50 เพื่อการเปรียบเทียบ
scenarios_to_evaluate = [{"ratio": "Baseline (0.50)", "threshold": 0.50, "cost_fn": 1, "cost_fp": 1}] + \
                        [{"ratio": f"Optimal under {scen['ratio']}", "threshold": optimal_thresholds[scen['ratio']], "cost_fn": scen['cost_fn'], "cost_fp": scen['cost_fp']} for scen in cost_scenarios]

for run in scenarios_to_evaluate:
    th = run["threshold"]
    cfn = run["cost_fn"]
    cfp = run["cost_fp"]

    test_preds = (test_probs >= th).astype(int)
    tn, fp, fn, tp = confusion_matrix(y_test, test_preds).ravel()

    expected_cost_test = (cfn * fn) + (cfp * fp)

    report_dict = classification_report(y_test, test_preds, output_dict=True, zero_division=0)
    precision_c1 = report_dict['1']['precision']
    recall_c1 = report_dict['1']['recall']
    f1_c1 = report_dict['1']['f1-score']

    test_evaluation_results.append({
        "Scenario": run["ratio"],
        "Decision Threshold": f"{th:.2f}",
        "Confusion Matrix (TN/FP/FN/TP)": f"{tn}/{fp}/{fn}/{tp}",
        "Precision (Class 1)": f"{precision_c1:.2f}",
        "Recall (Class 1)": f"{recall_c1:.2f}",
        "F1-Score (Class 1)": f"{f1_c1:.2f}",
        "Test Expected Cost": f"{expected_cost_test:.2f} (Ratio {cfn}:{cfp})"
    })

# จัดทำตารางเปรียบเทียบผลลัพธ์และจัดแสดง
df_test_evaluation = pd.DataFrame(test_evaluation_results)
print(f"\nHoldout Test Set Average Precision (AP) [คำนวณผ่านพื้นที่ใต้กราฟ Precision-Recall]: {ap_score_test:.4f}\n")
display(df_test_evaluation)
```

### ⚠️ ข้อจำกัดและข้อควรระวังในการนำไปประยุกต์ใช้เชิงธุรกิจ (Operational Constraints & Dataset Limitations)

1. **ลักษณะจำลองของชุดข้อมูล (Fictional / Synthetic Dataset):** ชุดข้อมูล IBM HR Attrition นี้ถูกจัดทำและเผยแพร่ในลักษณะจำลอง (Synthetic/Benchmark Dataset) เพื่อการทดลองทางวิทยาศาสตร์ข้อมูลและทดสอบแนวคิดเท่านั้น ไม่ได้สะท้อนถึงข้อมูลพฤติกรรมการทำงาน ทัศนคติ และวัฒนธรรมจริงของผู้ปฏิบัติงานในปัจจุบัน
2. **การนำผลการทดสอบไปใช้งานจริง:** ผลสัมฤทธิ์หรือทัศนะเชิงตัวเลขที่ปรากฏจากตัวแบบนี้ **ยังไม่สามารถนำไปยืนยันประสิทธิภาพในการตรวจจับพนักงานกลุ่มเสี่ยงในองค์กรจริงได้โดยทันที** จำเป็นอย่างยิ่งที่จะต้องมีการรวบรวม ทดสอบ และทำนายพฤติกรรมบนโครงสร้างข้อมูลที่เป็นปัจจุบันของพนักงานจริงในบริษัทแต่ละแห่งเสียก่อน
3. **การประนีประนอมประสิทธิภาพ (Trade-off):** การเลือกใช้อัตราส่วนต้นทุนที่ให้ความสำคัญกับการลด False Negative (การละเลยคนที่จะออกจริง) จะกระตุ้นให้ระบบมีค่า Recall สูงขึ้น แต่ต้องแลกมาด้วยการเกิด False Positive (การทักเตือนผิดคน) ที่เพิ่มพูนขึ้นตามลำดับ ซึ่งส่งผลกระทบโดยตรงต่อทรัพยากรการเข้าดูแลพนักงานและชั่วโมงทำงานของฝ่าย HR

### 📊 บทวิเคราะห์เปรียบเทียบเชิงวิชาการ (Academic Evaluation & Performance Trade-offs)

จากการทดสอบกลยุทธ์การจัดการปัญหาข้อมูลไม่สมดุล (Class Imbalance) ทั้ง 6 รูปแบบบนตัวแบบพยากรณ์ Random Forest ภายใต้เกณฑ์การตัดสินใจเริ่มต้นที่ระดับ **Decision Threshold = 0.50** มีประเด็นสำคัญที่สามารถอภิปรายเชิงสถิติศาสตร์ได้ดังนี้:

#### 1. การประเมินสมรรถนะภาพรวมด้วย Average Precision (AP)
* **Combined (SMOTE + Tomek) RF** มีความสามารถในการจัดอันดับคะแนนความเสี่ยง (Risk Ranking) ดีที่สุด โดยได้ค่า **AP สูงที่สุดที่ 0.5023** ตามด้วย **Oversampling (SMOTE) RF (AP = 0.4891)**
* การใช้เทคนิคไฮบริดอย่าง **SMOTE + Tomek** ช่วยปรับปรุงขอบเขตการตัดสินใจของข้อมูลคลาสย่อย (Minority Class) ได้ดีขึ้น อย่างไรก็ตาม พึงตระหนักว่าเทคนิคดังกล่าวช่วยบรรเทาแต่ไม่ได้ขจัดปัญหาความคาบเกี่ยวของคลาส (Class Overlap) ในมิติสูงออกไปได้อย่างสมบูรณ์ ส่งผลให้ยังมีพนักงานที่ไม่ได้ลาออกจริงบางส่วนถูกทำนายผิดพลาดอยู่

#### 2. ข้อพิจารณาในการเลือกใช้โมเดลและการแลกเปลี่ยนประสิทธิภาพ (Precision-Recall Trade-offs)
* **Undersampling (Full RUS 1:1) RF:** ให้ค่า **Recall สูงสุดอย่างมีนัยสำคัญ (0.72)** (ตรวจจับพนักงานที่ต้องการลาออกจริงได้ถึง 34 จาก 47 คน) แต่ต้องแลกมาด้วยค่า **Precision ที่ต่ำที่สุด (0.37)** และอัตราการเตือนพลาด (False Positives) ที่สูงมากถึง 59 คน ซึ่งทำให้ทรัพยากรการดูแลของ HR อาจถูกกระจายไปอย่างไม่ทั่วถึง
* **Combined (SMOTE + Tomek) RF & Oversampling (SMOTE) RF:** มอบจุดสมดุลที่ดีกว่า (Balanced Trade-off) โดยได้ค่า F1-Score ในระดับที่ยอมรับได้ (0.30 - 0.33) และสามารถควบคุมค่าความแม่นยำ (Precision) ให้อยู่ในระดับสูง (0.64 - 0.71) ซึ่งลดการแจ้งเตือนพนักงานผิดกลุ่มได้อย่างมีนัยสำคัญ
* **Baseline RF & Cost-Sensitive RF:** แม้จะมีความแม่นยำสูงเมื่อทำการแจ้งเตือน แต่กลับมีข้อจำกัดอย่างรุนแรงในเรื่องความไว (Recall ต่ำมากเพียง 0.11) ซึ่งส่งผลให้องค์กรไม่สามารถตรวจจับพนักงานกลุ่มเสี่ยงส่วนใหญ่ได้ทันท่วงทีก่อนเกิดการลาออกจริง

### 🧪 การเปรียบเทียบ Cost-Sensitive Decision Threshold ของ Random Forest 3 แบบ (Undersampling Only)

ส่วนการวิเคราะห์นี้ได้รับการออกแบบขึ้นใหม่ตามเงื่อนไขที่เข้มงวดเพื่อเปรียบเทียบประสิทธิภาพของโมเดล **Random Forest** 3 รูปแบบ:
1. **Baseline RF** (ไม่มีการ Resample และไม่มี Class Weight)
2. **Partial Undersampling RF (3:1)** (ลดสัดส่วน No ต่อ Yes ในข้อมูลฝึกให้เหลือ 3:1)
3. **Full Undersampling RF (1:1)** (ลดสัดส่วน No ต่อ Yes ในข้อมูลฝึกให้เหลือ 1:1)

**ข้อกำหนดที่คุมเข้ม:**
* ปราศจากการทำ Oversampling (ไม่มี SMOTE/SMOTENC/ADASYN/Combined) ทุกกรณี
* ไม่มีการใช้งาน `class_weight` (กำหนดเป็น `None` ใน RF เสมอ)
* กระบวนการ Resampling เกิดขึ้นเฉพาะใน Training Fold / Train Set เท่านั้น เพื่อควบคุม Data Leakage โดยสมบูรณ์
* การคำนวณและเปรียบเทียบจะใช้ชุดทดสอบ Holdout Test Set ชุดเดิมเดียวร่วมกัน

```python
import numpy as np
import pandas as pd
from sklearn.model_selection import StratifiedKFold
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import confusion_matrix, average_precision_score, classification_report
from imblearn.under_sampling import RandomUnderSampler
from imblearn.pipeline import Pipeline as ImbPipeline

# 1. นิยามสมมติฐาน Cost Scenario (FN:FP)
cost_ratios = [
    {"ratio": "1:1", "cost_fn": 1.0, "cost_fp": 1.0},
    {"ratio": "2:1", "cost_fn": 2.0, "cost_fp": 1.0},
    {"ratio": "5:1", "cost_fn": 5.0, "cost_fp": 1.0},
    {"ratio": "10:1", "cost_fn": 10.0, "cost_fp": 1.0}
]

# 2. จัดโครงสร้างสำหรับการจัดเก็บค่าเพื่อคำนวณ CV
skf_cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
threshold_grid = np.linspace(0.01, 0.99, 99)

# โมเดลและวิธีการที่จะประเมิน
strategies = ["Baseline RF", "Partial Undersampling RF (3:1)", "Full Undersampling RF (1:1)"]

# พจนานุกรมเพื่อเก็บ Out-of-fold predicted probabilities สำหรับแต่ละโมเดล
oof_probs = {strategy: np.zeros(len(y_train)) for strategy in strategies}
oof_indices = np.zeros(len(y_train), dtype=int)

# ดำเนินการ Cross-Validation เพื่อดึงคะแนนทำนายมาเลือก Threshold ปราศจาก Data Leakage
for fold, (train_idx, val_idx) in enumerate(skf_cv.split(X_train, y_train)):
    X_tr_fold, X_val_fold = X_train.iloc[train_idx], X_train.iloc[val_idx]
    y_tr_fold, y_val_fold = y_train.iloc[train_idx], y_train.iloc[val_idx]

    minority_count_fold = (y_tr_fold == 1).sum()

    for strategy in strategies:
        if strategy == "Baseline RF":
            pipeline = ImbPipeline([
                ('rf', RandomForestClassifier(random_state=42, class_weight=None))
            ])
        elif strategy == "Partial Undersampling RF (3:1)":
            strategy_ratio = {0: minority_count_fold * 3, 1: minority_count_fold}
            pipeline = ImbPipeline([
                ('under', RandomUnderSampler(sampling_strategy=strategy_ratio, random_state=42)),
                ('rf', RandomForestClassifier(random_state=42, class_weight=None))
            ])
        elif strategy == "Full Undersampling RF (1:1)":
            pipeline = ImbPipeline([
                ('under', RandomUnderSampler(sampling_strategy=1.0, random_state=42)),
                ('rf', RandomForestClassifier(random_state=42, class_weight=None))
            ])

        # เทรนโมเดลเฉพาะบน Train Fold
        pipeline.fit(X_tr_fold, y_tr_fold)
        # คาดการณ์ความน่าจะเป็นบน Validation Fold
        val_prob = pipeline.predict_proba(X_val_fold)[:, 1]
        oof_probs[strategy][val_idx] = val_prob

# ค้นหา Threshold ที่ให้ Expected Cost ต่ำที่สุดสำหรับแต่ละวิธีแยกตาม Cost Ratio
selected_thresholds = {strategy: {} for strategy in strategies}

for strategy in strategies:
    val_probs = oof_probs[strategy]
    for scen in cost_ratios:
        ratio_name = scen["ratio"]
        cfn = scen["cost_fn"]
        cfp = scen["cost_fp"]

        costs = []
        for th in threshold_grid:
            preds = (val_probs >= th).astype(int)
            tn, fp, fn, tp = confusion_matrix(y_train, preds).ravel()
            expected_cost = (cfn * fn) + (cfp * fp)
            costs.append(expected_cost)

        best_idx = np.argmin(costs)
        selected_thresholds[strategy][ratio_name] = threshold_grid[best_idx]

print("=== [Selected Optimal Thresholds from Validation (CV)] ===")
for strategy in strategies:
    print(f"📌 {strategy}:")
    for ratio, th in selected_thresholds[strategy].items():
         print(f"  - Cost Ratio {ratio} -> Selected Threshold: {th:.2f}")
```

```python
# ฝึกโมเดลตัวเต็มบนชุดข้อมูลฝึกอบรม (Train Set) ทั้งหมดเพื่อทดสอบ Holdout Test Set เพียงครั้งเดียว
final_models = {}
minority_count_total = (y_train == 1).sum()

for strategy in strategies:
    if strategy == "Baseline RF":
        pipeline = ImbPipeline([
            ('rf', RandomForestClassifier(random_state=42, class_weight=None))
        ])
    elif strategy == "Partial Undersampling RF (3:1)":
        strategy_ratio = {0: minority_count_total * 3, 1: minority_count_total}
        pipeline = ImbPipeline([
            ('under', RandomUnderSampler(sampling_strategy=strategy_ratio, random_state=42)),
            ('rf', RandomForestClassifier(random_state=42, class_weight=None))
        ])
    elif strategy == "Full Undersampling RF (1:1)":
        pipeline = ImbPipeline([
            ('under', RandomUnderSampler(sampling_strategy=1.0, random_state=42)),
            ('rf', RandomForestClassifier(random_state=42, class_weight=None))
        ])

    pipeline.fit(X_train, y_train)
    final_models[strategy] = pipeline

# ทำนายผลบนชุดทดสอบ Holdout Test Set (y_test ประกอบไปด้วยคลาส 0 = 247, คลาส 1 = 47, รวม = 294)
test_predictions = {}
for strategy in strategies:
    probs = final_models[strategy].predict_proba(X_test)[:, 1]
    test_predictions[strategy] = probs
```

```python
# --- ตาราง A: ความสามารถในการจัดอันดับคะแนนของแต่ละวิธี ---
table_a_data = []
for strategy in strategies:
    test_probs = test_predictions[strategy]
    # คำนวณ Average Precision (AP) ผ่านพื้นที่ใต้กราฟ Precision-Recall บนชุดข้อมูลทดสอบ
    ap_score = average_precision_score(y_test, test_probs)
    table_a_data.append({
        "วิธีของโมเดล (Model Strategy)": strategy,
        "Average Precision (AP) on Test Set": f"{ap_score:.4f}",
        "วิธีคำนวณ (Calculation Note)": "คำนวณค่าเฉลี่ยความถูกต้องแบบถ่วงน้ำหนักตามสัดส่วน Recall (scikit-learn average_precision_score) บน Holdout Test Set"
    })

df_table_a = pd.DataFrame(table_a_data)
print("=== ตาราง A: ความสามารถในการจัดอันดับคะแนนของแต่ละวิธี (Ranking Performance) ===")
display(df_table_a)
```

```python
# --- ตาราง B: ผลบน test set สำหรับแต่ละวิธีและแต่ละ cost ratio ---
table_b_rows = []

for strategy in strategies:
    test_probs = test_predictions[strategy]

    for scen in cost_ratios:
        ratio_name = scen["ratio"]
        cfn = scen["cost_fn"]
        cfp = scen["cost_fp"]

        # ระดับ Threshold ที่เลือกมาจาก CV และแบบดั้งเดิม 0.50
        selected_th = selected_thresholds[strategy][ratio_name]
        evals = [
            {"type": "Threshold 0.50", "th_value": 0.50},
            {"type": f"Optimal CV ({selected_th:.2f})", "th_value": selected_th}
        ]

        for item in evals:
            th_val = item["th_value"]
            preds = (test_probs >= th_val).astype(int)
            cm = confusion_matrix(y_test, preds)
            tn, fp, fn, tp = cm.ravel()

            # ตรวจสอบขนาดรวมของชุดข้อมูลทดสอบ
            assert (tn + fp + fn + tp) == len(y_test), "ขนาดข้อมูลรวมใน Confusion Matrix ไม่สอดคล้องกับขนาดของชุดข้อมูลทดสอบ"

            report_dict = classification_report(y_test, preds, output_dict=True, zero_division=0)
            p_c1 = report_dict['1']['precision']
            r_c1 = report_dict['1']['recall']
            f1_c1 = report_dict['1']['f1-score']

            # คำนวณ Expected Cost จากสูตร (Cost_FN * FN) + (Cost_FP * FP)
            total_cost = (cfn * fn) + (cfp * fp)
            cost_per_100 = (total_cost / len(y_test)) * 100

            table_b_rows.append({
                "วิธีของโมเดล": strategy,
                "Cost Ratio FN:FP": ratio_name,
                "ระดับเกณฑ์ตัดสินใจ": item["type"],
                "Confusion Matrix (TN/FP/FN/TP)": f"{tn}/{fp}/{fn}/{tp}",
                "Precision (Class 1)": f"{p_c1:.2f}",
                "Recall (Class 1)": f"{r_c1:.2f}",
                "F1-Score (Class 1)": f"{f1_c1:.2f}",
                "Expected Cost รวม": f"{total_cost:.2f}",
                "Expected Cost ต่อ 100 คน": f"{cost_per_100:.2f}"
            })

df_table_b = pd.DataFrame(table_b_rows)
print("=== ตาราง B: ผลบน Test Set สำหรับแต่ละวิธีและแต่ละ Cost Ratio ===")
pd.set_option('display.max_rows', 50)
display(df_table_b)
```

### 🧪 การประเมินผลความสม่ำเสมอด้วย Repeated Stratified Cross-Validation ร่วมกับ Nested Threshold Optimization

เพื่อยืนยันและประเมินประสิทธิภาพเชิงสถิติของกลยุทธ์การจัดการ Imbalance ทั้ง 3 รูปแบบ (Baseline, Partial 3:1, และ Full 1:1) โดยไม่เกิดปัญหาความผันผวนจากการแบ่งข้อมูลแบบรอบเดียว (Single Split) เราจะดำเนินการรัน **Repeated Stratified Cross-Validation (5 Folds, 5 Repeats)** บนชุดข้อมูลสอน `X_train` และ `y_train` ดั้งเดิม โดยละเว้นข้อมูล Holdout Test Set (`X_test`, `y_test`) ไว้สำหรับทดสอบขั้นสุดท้ายเท่านั้น

#### ข้อกำหนดในการทดลองอย่างเข้มงวด:
1. **Nested CV for Threshold Optimization:** สำหรับแต่ละ Outer Fold เราจะแบ่งข้อมูลเป็น Inner Fold เพื่อค้นหา Optimal Threshold ที่ให้ค่า Expected Cost ต่ำที่สุดของแต่ละ Cost Ratio (1:1, 2:1, 5:1, 10:1) แล้วนำ Threshold นั้นมาประเมินค่าใช้จ่ายบน Outer Validation Fold จริง
2. **No Data Leakage & Resampling Control:** กระบวนการ Undersampling (ด้วย `RandomUnderSampler`) จะเกิดขึ้นบนข้อมูลสอนในแต่ละ Fold เท่านั้น โดยคลาส Validation Fold ของทั้ง Inner และ Outer จะคงอัตราส่วนจริงไว้ไม่มีการแปลง และไม่มีการใช้ SMOTE, Oversampling หรือ `class_weight` ใดๆ ทั้งสิ้น
3. **Consistency:** ทุกโมเดลและทุกรอบการสุ่มจะรันภายใต้โครงสร้าง Fold และการสุ่มชุดเดียวกัน (`random_state=42` สม่ำเสมอ)

```python
import numpy as np
import pandas as pd
from sklearn.model_selection import RepeatedStratifiedKFold, StratifiedKFold
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import confusion_matrix, average_precision_score, precision_recall_fscore_support
from imblearn.under_sampling import RandomUnderSampler
from imblearn.pipeline import Pipeline as ImbPipeline

# 1. กำหนดพารามิเตอร์เริ่มต้นและ Cost Ratios สำหรับประเมิน
cost_scenarios = [
    {"ratio": "1:1", "cost_fn": 1.0, "cost_fp": 1.0},
    {"ratio": "2:1", "cost_fn": 2.0, "cost_fp": 1.0},
    {"ratio": "5:1", "cost_fn": 5.0, "cost_fp": 1.0},
    {"ratio": "10:1", "cost_fn": 10.0, "cost_fp": 1.0}
]

# กำหนดโมเดลเป้าหมายทั้ง 3 กลยุทธ์
strategies = [
    "Baseline RF",
    "Partial Undersampling RF (3:1)",
    "Full Undersampling RF (1:1)"
]

# 2. เตรียมโครงสร้าง Repeated Stratified K-Fold (5 folds, 5 repeats)
rskf = RepeatedStratifiedKFold(n_splits=5, n_repeats=5, random_state=42)
threshold_grid = np.linspace(0.01, 0.99, 99)

# ตัวแปรเก็บผลลัพธ์ Outer Loop Metrics (ประเมินที่ default threshold = 0.50)
outer_metrics_results = {strat: [] for strat in strategies}

# ตัวแปรเก็บผลลัพธ์ Nested Threshold & Expected Costs แยกตามกลยุทธ์และอัตราส่วนต้นทุน
nested_cost_results = {strat: {scen["ratio"]: [] for scen in cost_scenarios} for strat in strategies}

# เริ่มต้นวนรอบกระบวนการ Repeated CV (Outer Loop)
for run_idx, (outer_train_idx, outer_val_idx) in enumerate(rskf.split(X_train, y_train)):
    X_ot_train, X_ot_val = X_train.iloc[outer_train_idx], X_train.iloc[outer_val_idx]
    y_ot_train, y_ot_val = y_train.iloc[outer_train_idx], y_train.iloc[outer_val_idx]

    # ดึงจำนวนคลาสกลุ่มน้อย (Class 1) ใน Outer Train
    minority_count_ot = (y_ot_train == 1).sum()

    # สร้างโมเดลปลายทางของแต่ละกลยุทธ์สำหรับ Outer Loop
    models_ot = {}
    for strat in strategies:
        if strat == "Baseline RF":
            pipeline = ImbPipeline([('rf', RandomForestClassifier(random_state=42, class_weight=None))])
        elif strat == "Partial Undersampling RF (3:1)":
            strategy_ratio = {0: minority_count_ot * 3, 1: minority_count_ot}
            pipeline = ImbPipeline([
                ('under', RandomUnderSampler(sampling_strategy=strategy_ratio, random_state=42)),
                ('rf', RandomForestClassifier(random_state=42, class_weight=None))
            ])
        elif strat == "Full Undersampling RF (1:1)":
            pipeline = ImbPipeline([
                ('under', RandomUnderSampler(sampling_strategy=1.0, random_state=42)),
                ('rf', RandomForestClassifier(random_state=42, class_weight=None))
            ])
        pipeline.fit(X_ot_train, y_ot_train)
        models_ot[strat] = pipeline

    # A) ประเมินผลประสิทธิภาพมาตรฐานของแบบจำลองที่ Threshold = 0.50 และคำนวณ AP
    for strat in strategies:
        pipeline = models_ot[strat]
        val_probs = pipeline.predict_proba(X_ot_val)[:, 1]
        val_preds = (val_probs >= 0.50).astype(int)

        ap_score = average_precision_score(y_ot_val, val_probs)
        prec, rec, f1, _ = precision_recall_fscore_support(y_ot_val, val_preds, average='binary', pos_label=1, zero_division=0)
        cm = confusion_matrix(y_ot_val, val_preds, labels=[0, 1])
        tn, fp, fn, tp = cm.ravel() if cm.size == 4 else (0, 0, 0, 0)

        outer_metrics_results[strat].append({
            'ap': ap_score,
            'precision': prec,
            'recall': rec,
            'f1': f1,
            'tn': tn, 'fp': fp, 'fn': fn, 'tp': tp
        })

    # B) ค้นหาจุดตัดด้วย Inner Stratified 3-Fold CV เพื่อป้องกัน Optimistic Bias (Nested Threshold Optimization)
    inner_skf = StratifiedKFold(n_splits=3, shuffle=True, random_state=42)
    inner_oof_probs = {strat: np.zeros(len(y_ot_train)) for strat in strategies}

    for inner_train_idx, inner_val_idx in inner_skf.split(X_ot_train, y_ot_train):
        X_in_train, X_in_val = X_ot_train.iloc[inner_train_idx], X_ot_train.iloc[inner_val_idx]
        y_in_train, y_in_val = y_ot_train.iloc[inner_train_idx], y_ot_train.iloc[inner_val_idx]

        minority_count_in = (y_in_train == 1).sum()

        for strat in strategies:
            if strat == "Baseline RF":
                in_pipeline = ImbPipeline([('rf', RandomForestClassifier(random_state=42, class_weight=None))])
            elif strat == "Partial Undersampling RF (3:1)":
                strategy_ratio_in = {0: minority_count_in * 3, 1: minority_count_in}
                in_pipeline = ImbPipeline([
                    ('under', RandomUnderSampler(sampling_strategy=strategy_ratio_in, random_state=42)),
                    ('rf', RandomForestClassifier(random_state=42, class_weight=None))
                ])
            elif strat == "Full Undersampling RF (1:1)":
                in_pipeline = ImbPipeline([
                    ('under', RandomUnderSampler(sampling_strategy=1.0, random_state=42)),
                    ('rf', RandomForestClassifier(random_state=42, class_weight=None))
                ])
            in_pipeline.fit(X_in_train, y_in_train)
            inner_oof_probs[strat][inner_val_idx] = in_pipeline.predict_proba(X_in_val)[:, 1]

    # ค้นหาค่าเกณฑ์ที่เหมาะสมที่สุด (Optimal Inner Threshold) และนำมาประเมินบน Outer Validation Set ชุดจริง
    for strat in strategies:
        in_probs = inner_oof_probs[strat]
        ot_pipeline = models_ot[strat]
        ot_probs = ot_pipeline.predict_proba(X_ot_val)[:, 1]

        for scen in cost_scenarios:
            ratio_name = scen["ratio"]
            cfn = scen["cost_fn"]
            cfp = scen["cost_fp"]

            # คำนวณ Expected Cost บน Inner OOF เพื่อเจาะหา Threshold ที่ดีที่สุด
            inner_costs = []
            for th in threshold_grid:
                in_preds = (in_probs >= th).astype(int)
                tn_in, fp_in, fn_in, tp_in = confusion_matrix(y_ot_train, in_preds, labels=[0, 1]).ravel()
                inner_costs.append((cfn * fn_in) + (cfp * fp_in))

            best_th = threshold_grid[np.argmin(inner_costs)]

            # นำเกณฑ์ที่เลือก (best_th) มาประเมินแบบไม่มีส่วนได้เสียบน Outer Validation Set ชุดจริง
            ot_preds = (ot_probs >= best_th).astype(int)
            tn_ot, fp_ot, fn_ot, tp_ot = confusion_matrix(y_ot_val, ot_preds, labels=[0, 1]).ravel()
            final_ot_expected_cost = (cfn * fn_ot) + (cfp * fp_ot)

            nested_cost_results[strat][ratio_name].append({
                'threshold': best_th,
                'expected_cost': final_ot_expected_cost
            })

print("=== Repeated Stratified CV Completed Successfully! ===")
```

```python
# 3. การประมวลและแปลงค่าเฉลี่ยสรุปผลแบบตารางเพื่อความแม่นยำทางสถิติศาสตร์
summary_metrics = []

for strat in strategies:
    df_strat_m = pd.DataFrame(outer_metrics_results[strat])
    summary_metrics.append({
        'Model Strategy': strat,
        'Mean AP (± SD)': f"{df_strat_m['ap'].mean():.4f} (± {df_strat_m['ap'].std():.4f})",
        'Precision (Class 1) (± SD)': f"{df_strat_m['precision'].mean():.2f} (± {df_strat_m['precision'].std():.2f})",
        'Recall (Class 1) (± SD)': f"{df_strat_m['recall'].mean():.2f} (± {df_strat_m['recall'].std():.2f})",
        'F1-Score (Class 1) (± SD)': f"{df_strat_m['f1'].mean():.2f} (± {df_strat_m['f1'].std():.2f})",
        'Confusion Matrix (Mean TN/FP/FN/TP)': f"{df_strat_m['tn'].mean():.1f} / {df_strat_m['fp'].mean():.1f} / {df_strat_m['fn'].mean():.1f} / {df_strat_m['tp'].mean():.1f}"
    })

df_summary_m = pd.DataFrame(summary_metrics)
print("=== ตารางที่ 1: สรุปผลประสิทธิภาพโมเดลเฉลี่ยจากการทำ 5-Fold 5-Repeat CV (ที่ Threshold = 0.50) ===")
display(df_summary_m)

# 4. การประมวลและสรุปผล Nested Cost-Sensitive Optimization
summary_nested_costs = []
for strat in strategies:
    for scen in cost_scenarios:
        ratio_name = scen["ratio"]
        results_list = nested_cost_results[strat][ratio_name]

        ths = [res['threshold'] for res in results_list]
        costs = [res['expected_cost'] for res in results_list]

        summary_nested_costs.append({
            'Model Strategy': strat,
            'Cost Ratio FN:FP (Assumption)': ratio_name,
            'Mean Optimal Threshold (± SD)': f"{np.mean(ths):.2f} (± {np.std(ths):.2f})",
            'Mean Expected Cost (± SD)': f"{np.mean(costs):.2f} (± {np.std(costs):.2f})",
            'Expected Cost per 100 Employees': f"{(np.mean(costs) / len(y_ot_val) * 100):.2f}"
        })

df_summary_nested = pd.DataFrame(summary_nested_costs)
print("\n=== ตารางที่ 2: ผลประเมินแบบไร้ส่วนได้เสียด้วย Nested Decision Threshold Optimization และ Expected Cost เฉลี่ย (ต่อ 1 Validation Fold) ===")
print("*หมายเหตุ: อัตราส่วนต้นทุน (Cost Ratios) เป็นสมมติฐานการวิเคราะห์ความไว (Sensitivity Analysis) เชิงตัวแบบเท่านั้น ไม่ใช่ค่าใช้จ่ายจริงขององค์กร")
display(df_summary_nested)
```

### 📊 บทวิเคราะห์เปรียบเทียบเชิงลึกหลังการตรวจสอบความสม่ำเสมอ (Detailed Trade-off Analysis & Key Takeaways)

จากการรันการทดลอง **Repeated Stratified CV** ร่วมกับระบบ **Nested Threshold Selection** พบข้อสังเกตและพฤติกรรมเชิงลึกของแต่ละวิธีการจัดการ Imbalance (Undersampling Only) ดังนี้:

#### 1. ความสามารถในการแยกแยะข้อมูลโดยไม่จำกัด Threshold (Average Precision: AP)
* **Baseline Random Forest (No Resampling)** ยังคงให้ค่าเฉลี่ย **Average Precision (AP)** สูงที่สุดอย่างสม่ำเสมอและมีความเสถียรที่สุดเมื่อเปลี่ยนกลุ่มแบ่งข้อมูล เนื่องจากตัวแบบสามารถเรียนรู้ความซับซ้อนของข้อมูลดิบได้อย่างสมบูรณ์ ปราศจากการทำลายสารสนเทศจากฝั่งคลาสส่วนมาก (No Information Loss)
* การทำ **Undersampling ทั้งสองระดับ** ส่งผลให้เกิดการกระจายคะแนนความเป็นไปได้ของผลทำนายที่บิดเบือนไปจากความถี่ประชากรจริง และมีค่าเบี่ยงเบนมาตรฐาน (SD) ของค่า AP ที่กว้างขึ้นเล็กน้อย สะท้อนถึงผลกระทบจากการสูญเสียข้อมูลจากการสุ่มตัดแถวทิ้ง

#### 2. ความคุ้มค่าทางเศรษฐศาสตร์ธุรกิจภายใต้อัตราส่วนต้นทุนแบบจำลอง (Precision-Recall Trade-offs & Expected Cost)
* **ในกรณีความต้องการจับสัญญาณพนักงานอย่างแม่นยำสูง (Cost Ratio 1:1, 2:1):** ตัวแบบ **Baseline RF** ให้ประสิทธิภาพในภาพรวม (F1-score & Precision) ที่เด่นชัดกว่า และประหยัดค่าใช้จ่ายเชิงรวมได้เนื่องจากมีจำนวน False Positives (ทักพนักงานผิดตัว) ที่ต่ำมาก
* **ในกรณีความต้องการหลีกเลี่ยงการลาออกเชิงรุกขั้นวิกฤต (Cost Ratio 5:1, 10:1):**
  * ตัวแบบ **Full Undersampling RF (1:1)** และ **Partial Undersampling RF (3:1)** ที่ผ่านการปรับค่า Threshold ให้ต่ำลงผ่าน Nested CV จะโดดเด่นอย่างยิ่งในการเพิ่มระดับ Recall (ตรวจพบพนักงานที่จะออกได้ครอบคลุมมากขึ้น)
  * อย่างไรก็ดี ต้องแลกมาด้วยการแบกรับค่า Precision ที่ต่ำลงมาก (False Positive บวมขึ้น) ซึ่งในเชิงปฏิบัติฝ่ายทรัพยากรบุคคล (HR) จะต้องมีงบประมาณและกำลังเจ้าหน้าที่ที่เพียงพอในการเข้าสืบค้นและทำกิจกรรมดูแลพนักงานกลุ่มใหญ่ที่ถูกตรวจพบโดยเครื่องมือชิ้นนี้
