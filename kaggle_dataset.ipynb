2. **การกำหนด Class Weights (ในตัวโมเดล)**

   * **Undersampling**: สุ่มลดจำนวนข้อมูลคลาสกลุ่มมาก (No) ลงให้เหมาะสม

### 🧱 การจัดเตรียมข้อมูลและการแบ่งข้อมูลเป็น Train และ Test set
ก่อนอื่นเราจะกำหนดตัวแปรต้น (Features) และตัวแปรเป้าหมาย (Target) รวมถึงแปลงตัวแปรเป้าหมาย `Attrition` ให้เป็นตัวเลข (Yes = 1, No = 0) จากนั้นใช้ฟังก์ชัน `train_test_split` โดยกำหนดพารามิเตอร์ `stratify=y` เพื่อให้มั่นใจว่าสัดส่วนของกลุ่มคนลาออก (Class Distribution) ในข้อมูล Train และ Test ยังคงเท่ากัน

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer
import pandas as pd
import numpy as np

# 1. แปลงตัวแปรเป้าหมาย (Target)
y = df['Attrition'].apply(lambda x: 1 if x == 'Yes' else 0)

# 2. คัดเลือก Features และลบคอลัมน์ที่ไม่ส่งผลเชิงทำนายออก
columns_to_drop = ['Attrition', 'EmployeeNumber', 'EmployeeCount', 'Over18', 'StandardHours']
X_raw = df.drop(columns=[col for col in columns_to_drop if col in df.columns])

# 3. แบ่งชุดข้อมูลดิบเป็น Train และ Test (stratify ป้องกันความไม่สมดุล)
X_train_raw, X_test_raw, y_train, y_test = train_test_split(
    X_raw, y, test_size=0.2, random_state=42, stratify=y
)

# 4. ระบุคอลัมน์ประเภทหมวดหมู่ (Categorical Features) เพื่อทำ One-Hot Encoding
categorical_cols = X_train_raw.select_dtypes(include=['object', 'category']).columns.tolist()
numeric_cols = X_train_raw.select_dtypes(exclude=['object', 'category']).columns.tolist()

# 5. ใช้ ColumnTransformer เพื่อแปลง OneHotEncoder โดยฝึกสอน (fit) เฉพาะใน Train Set เท่านั้น
preprocessor = ColumnTransformer(
    transformers=[
        ('cat', OneHotEncoder(drop='first', sparse_output=False, handle_unknown='ignore'), categorical_cols)
    ],
    remainder='passthrough'
)

# ฟิตข้อมูลและแปลงค่าเพื่อสร้าง Features สำหรับเทรนและทดสอบ
X_train_arr = preprocessor.fit_transform(X_train_raw)
X_test_arr = preprocessor.transform(X_test_raw)

# ดึงชื่อฟีเจอร์หลังจากการทำแปลงข้อมูล One-Hot
encoded_cat_features = preprocessor.named_transformers_['cat'].get_feature_names_out(categorical_cols).tolist()
all_feature_names = encoded_cat_features + numeric_cols

# แปลงผลลัพธ์กลับเป็น pandas DataFrame เพื่อความสะดวกในการใช้งานกับ SHAP และวิเคราะห์ต่อ
X_train = pd.DataFrame(X_train_arr, columns=all_feature_names, index=X_train_raw.index)
X_test = pd.DataFrame(X_test_arr, columns=all_feature_names, index=X_test_raw.index)

print(f"จำนวนคอลัมน์หลังทำแปลงแบบไร้ Data Leakage: {X_train.shape[1]}")
print(f"พารามิเตอร์ OneHot ถูกเรียนรู้เฉพาะใน Train set สำเร็จ!")
```

```python
from imblearn.over_sampling import SMOTE
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report, precision_recall_curve, auc
import pandas as pd

# 1. ทำการ Oversampling ด้วย SMOTE บนข้อมูลที่ผ่าน ColumnTransformer (Train Set เท่านั้น)
smote = SMOTE(random_state=42)
X_train_resampled, y_train_resampled = smote.fit_resample(X_train, y_train)

# 2. ฝึกโมเดล Random Forest ด้วยข้อมูลที่ป้องกัน Data Leakage และปรับสมดุลแล้ว
smote_rf = RandomForestClassifier(random_state=42)
smote_rf.fit(X_train_resampled, y_train_resampled)

# 3. ทำนายผลบน Test Set
y_pred_smote = smote_rf.predict(X_test)
y_probs_smote = smote_rf.predict_proba(X_test)[:, 1]

# 4. แสดงรายงานผลการประเมิน
print("=== SMOTE + Random Forest (Clean Pipeline) Classification Report ===")
print(classification_report(y_test, y_pred_smote, target_names=['No (0)', 'Yes (1)']))

precision_s, recall_s, _ = precision_recall_curve(y_test, y_probs_smote)
pr_auc_smote = auc(recall_s, precision_s)
print(f"PR-AUC Score (SMOTE - Leakage Free): {pr_auc_smote:.4f}")
```

```python
# 5. คัดเลือกกลุ่มพนักงานที่มีความเสี่ยงสูง (High Risk) ด้วยระดับเกณฑ์เดิม Threshold >= 50%
test_analysis_df = X_test.copy()
test_analysis_df['Actual_Attrition'] = y_test
test_analysis_df['Predicted_Probability'] = y_probs_smote
test_analysis_df['Predicted_Attrition'] = y_pred_smote

high_risk_employees = test_analysis_df[test_analysis_df['Predicted_Probability'] >= 0.5].sort_values(by='Predicted_Probability', ascending=False)
print(f"=== จำนวนพนักงานในกลุ่มเสี่ยงสูงหลังปรับปรุง Preprocessing: {high_risk_employees.shape[0]} คน ===")
```

```python
่from sklearn.metrics import classification_report, confusion_matrix, precision_recall_curve, auc
import pandas as pd

# ฟังก์ชันช่วยสรุปผลลัพธ์ของแต่ละโมเดลอย่างละเอียด
def evaluate_model_pipeline(model, name, X_t, y_t):
    preds = model.predict(X_t)
    probs = model.predict_proba(X_t)[:, 1]

    # Confusion Matrix
    cm = confusion_matrix(y_t, preds)
    tn, fp, fn, tp = cm.ravel()

    # Precision-Recall AUC
    precision, recall, _ = precision_recall_curve(y_t, probs)
    pr_auc_val = auc(recall, precision)

    # ดึงค่า Recall และ F1-score ของ Class 1
    report = classification_report(y_t, preds, output_dict=True)
    rec_1 = report['1']['recall']
    f1_1 = report['1']['f1-score']

    print(f"=== {name} ===")
    print(f"Confusion Matrix: TN={tn}, FP={fp}, FN={fn}, TP={tp}")
    print(f"Recall (Class 1): {rec_1:.2f}")
    print(f"F1-Score (Class 1): {f1_1:.2f}")
    print(f"PR-AUC Score: {pr_auc_val:.4f}")
    print("-"*40)
    return tn, fp, fn, tp, rec_1, f1_1, pr_auc_val

# ประเมินผล 3 โมเดลบนชุดข้อมูลที่ป้องกัน Leakage แล้ว
_ = evaluate_model_pipeline(baseline_rf, "1. Baseline RF", X_test, y_test)
_ = evaluate_model_pipeline(balanced_rf, "2. Balanced RF", X_test, y_test)
_ = evaluate_model_pipeline(smote_rf, "3. SMOTE + RF", X_test, y_test)

# ยืนยันจำนวนพนักงานที่มี Probability >= 0.50 สำหรับ SMOTE + RF
prob_smote = smote_rf.predict_proba(X_test)[:, 1]
risk_count = (prob_smote >= 0.50).sum()
print(f"จำนวนพนักงานกลุ่มเสี่ยงสูง (Prob >= 50%) จากโมเดล SMOTE+RF: {risk_count} คน")
```

```python
from sklearn.model_selection import train_test_split
import numpy as np

# 1. แปลงตัวแปรเป้าหมาย (Target) Attrition จาก Yes/No เป็น 1/0
y = df['Attrition'].apply(lambda x: 1 if x == 'Yes' else 0)

# 2. กำหนดตัวแปรต้น (Features) โดยลบ Attrition ออก
# หมายเหตุ: เราสามารถลบ EmployeeNumber (Identifier) และคอลัมน์ที่ไม่ส่งผลต่อการวิเคราะห์ออกไปด้วย
columns_to_drop = ['Attrition', 'EmployeeNumber', 'EmployeeCount', 'Over18', 'StandardHours']
X = df.drop(columns=[col for col in columns_to_drop if col in df.columns])

# 3. จัดทำ One-Hot Encoding สำหรับ Categorical Features ตัวอื่นๆ
X_encoded = pd.get_dummies(X, drop_first=True)

# 4. แบ่งข้อมูลเป็น Train Set (80%) และ Test Set (20%)
# ใช้ stratify=y เพื่อรักษาอัตราส่วนของ Class Imbalance ให้เท่ากันทั้งสองเซ็ต
X_train, X_test, y_train, y_test = train_test_split(
    X_encoded, y, test_size=0.2, random_state=42, stratify=y
)

print(f"จำนวนแถวข้อมูลสอน (Train Set): {X_train.shape[0]} แถว, จำนวน Features: {X_train.shape[1]}")
print(f"จำนวนแถวข้อมูลทดสอบ (Test Set): {X_test.shape[0]} แถว, จำนวน Features: {X_test.shape[1]}")
print("\nสัดส่วนคลาสใน Train Set:")
print(y_train.value_counts(normalize=True) * 100)
print("\nสัดส่วนคลาสใน Test Set:")
print(y_test.value_counts(normalize=True) * 100)
```

### 🌲 การสร้างและประเมิน Baseline Random Forest
สร้างตัวแบบทำนาย (Classifier) พื้นฐานโดยไม่ใช้กลไกการถ่วงน้ำหนักหรือปรับสมดุล เพื่อใช้เป็นเกณฑ์เปรียบเทียบหลัก

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report, precision_recall_curve, auc
import matplotlib.pyplot as plt

# 1. เทรนโมเดล Baseline Random Forest (ไม่กำหนด class_weight)
baseline_rf = RandomForestClassifier(random_state=42)
baseline_rf.fit(X_train, y_train)

# 2. ทำนายผลบน Test Set
y_pred = baseline_rf.predict(X_test)
y_probs = baseline_rf.predict_proba(X_test)[:, 1]

# 3. แสดงรายงานการจำแนกประเภท (Classification Report)
print("=== Baseline Random Forest Classification Report ===")
print(classification_report(y_test, y_pred, target_names=['No (0)', 'Yes (1)']))

# 4. คำนวณ Precision-Recall AUC (PR-AUC)
precision, recall, thresholds = precision_recall_curve(y_test, y_probs)
pr_auc = auc(recall, precision)
print(f"PR-AUC Score: {pr_auc:.4f}")
```

เราจะกำหนดค่าพารามิเตอร์ `class_weight='balanced'` ในโมเดล Random Forest เพื่อเพิ่มการเรียนรู้ให้กับคลาสที่ลาออก (Yes) และเปรียบเทียบค่าพารามิเตอร์ที่ได้ว่าพัฒนาขึ้นจาก Baseline หรือไม่

```python
# 1. สร้างและฝึกโมเดลด้วย class_weight='balanced'
balanced_rf = RandomForestClassifier(class_weight='balanced', random_state=42)
balanced_rf.fit(X_train, y_train)

# 2. ทำนายผล
y_pred_balanced = balanced_rf.predict(X_test)
y_probs_balanced = balanced_rf.predict_proba(X_test)[:, 1]

# 3. แสดงรายงานการประเมิน
print("=== Balanced Random Forest Classification Report ===")
print(classification_report(y_test, y_pred_balanced, target_names=['No (0)', 'Yes (1)']))

# 4. คำนวณ Precision-Recall AUC (PR-AUC)
precision_b, recall_b, _ = precision_recall_curve(y_test, y_probs_balanced)
pr_auc_balanced = auc(recall_b, precision_b)
print(f"PR-AUC Score (Balanced): {pr_auc_balanced:.4f}")
```

### 🧪 การแก้ปัญหาโดยใช้ SMOTE (Synthetic Minority Over-sampling Technique)
เราจะใช้ไลบรารี `imblearn` เพื่อทำ Oversampling เฉพาะในชุดข้อมูลสอน (Train Set) เท่านั้น (เพื่อป้องกันปัญหา Data Leakage) แล้วนำข้อมูลที่สมบูรณ์ไปเทรนโมเดล Random Forest

```python
# ติดตั้ง imbalanced-learn หากยังไม่มีในสภาพแวดล้อม
!pip install -q imbalanced-learn

from imblearn.over_sampling import SMOTE

# 1. ทำการ Oversampling ด้วย SMOTE บนข้อมูล Train Set เท่านั้น
smote = SMOTE(random_state=42)
X_train_resampled, y_train_resampled = smote.fit_resample(X_train, y_train)

print("=== หลังทำ SMOTE ===")
print(f"จำนวนแถวข้อมูลสอนหลังทำ SMOTE: {X_train_resampled.shape[0]} แถว")
print("สัดส่วนคลาสหลังทำ SMOTE:")
print(pd.Series(y_train_resampled).value_counts(normalize=True) * 100)

# 2. ฝึกโมเดล Random Forest ด้วยข้อมูลที่ปรับสมดุลแล้ว
smote_rf = RandomForestClassifier(random_state=42)
smote_rf.fit(X_train_resampled, y_train_resampled)

# 3. ทำนายผล
y_pred_smote = smote_rf.predict(X_test)
y_probs_smote = smote_rf.predict_proba(X_test)[:, 1]

# 4. แสดงรายงานผล
print("\n=== SMOTE + Random Forest Classification Report ===")
print(classification_report(y_test, y_pred_smote, target_names=['No (0)', 'Yes (1)']))

# 5. คำนวณ PR-AUC
precision_s, recall_s, _ = precision_recall_curve(y_test, y_probs_smote)
pr_auc_smote = auc(recall_s, precision_s)
print(f"PR-AUC Score (SMOTE): {pr_auc_smote:.4f}")
```

### 📋 การสร้าง Rule-based Recommendations สำหรับป้องกันการลาออกของพนักงานกลุ่มเสี่ยง
เราจะกำหนดเกณฑ์และกฎ (Rules) อัตโนมัติโดยอิงจากปัจจัยสำคัญที่มักส่งผลต่อการตัดสินใจลาออก (เช่น รายได้ต่ำ, ทำโอเวอร์ไทม์หนัก, หรือต้องเดินทางไกล) เพื่อเสนอแนะแนวทางแก้ไขปัญหาที่เจาะจงรายบุคคลให้กับฝ่ายบุคคล (HR)

```python
def generate_hr_recommendation(row):
    recommendations = []

    # กฎข้อที่ 1: การทำโอเวอร์ไทม์ (OverTime)
    if 'OverTime_Yes' in row and row['OverTime_Yes'] == 1:
        recommendations.append("🔴 ลดภาระงานนอกเวลา (Overtime Redesign): ตรวจสอบการทำโอเวอร์ไทม์ และจัดสรรงานหรือปรับแผนงานเพื่อลดภาระงานหนัก")

    # กฎข้อที่ 2: รายได้ต่อเดือนต่ำกว่าเกณฑ์เฉลี่ยพนักงานทั่วไป
    if row['MonthlyIncome'] < 3000:
        recommendations.append("💵 ทบทวนโครงสร้างรายได้ (Compensation Review): พิจารณาปรับปรุงฐานเงินเดือนขั้นต่ำ หรือเพิ่มเบี้ยเลี้ยงพิเศษ")

    # กฎข้อที่ 3: เดินทางไกล
    if row['DistanceFromHome'] > 15:
        recommendations.append("🚗 นโยบายการทำงานยืดหยุ่น (Flexible Work): เสนอทางเลือกในการทำงานจากบ้าน (Work from Home) หรือความยืดหยุ่นในการเดินทาง")

    # กฎข้อที่ 4: ความพึงพอใจสิ่งแวดล้อมต่ำ
    if row['EnvironmentSatisfaction'] <= 2:
        recommendations.append("🏢 ปรับปรุงความสัมพันธ์และบรรยากาศ (Workplace Environment Improvement): พูดคุยเปิดอกเพื่อแก้ไขปัญหาในทีมหรือสภาพแวดล้อม")

    if len(recommendations) == 0:
        recommendations.append("🟢 พูดคุยและติดตาม (Career Development Check-in): ร่วมพูดคุยเพื่อวางแผนความก้าวหน้าในอาชีพ")

    return "\n".join([f"{i+1}. {rec}" for i, rec in enumerate(recommendations)])

# นำกฎที่สร้างขึ้นมาใช้กับพนักงานกลุ่มเสี่ยงสูง (High Risk)
risk_report = high_risk_employees.copy()
risk_report['Actionable_Recommendations'] = risk_report.apply(generate_hr_recommendation, axis=1)

# แสดงรายงานแนะนำการแก้ไขปัญหาของพนักงานที่มีความเสี่ยงสูงที่สุด 3 อันดับแรก
print("=== รายงานคำแนะนำในการดูแลรักษาพนักงานกลุ่มเสี่ยงสูงสุด (Top 3 Candidates) ===")
for idx, row in risk_report.head(3).iterrows():
    print(f"\nรหัสพนักงาน (Index): {idx}")
    print(f"- ความเสี่ยงในการลาออก: {row['Predicted_Probability']*100:.1f}%")
    print(f"- อายุ: {row['Age']} ปี | รายได้ต่อเดือน: ${row['MonthlyIncome']:.0f} | ระยะทางบ้าน: {row['DistanceFromHome']} ไมล์")
    print(f"- คำแนะนำสำหรับฝ่าย HR:\n{row['Actionable_Recommendations']}")
    print("-" * 80)
```

```python
import numpy as np
import pandas as pd
import shap
import matplotlib.pyplot as plt

def verify_individual_risk_path(employee_idx, model, X_test_df, y_test_series, risk_report_df, shap_explainer):
    """
    ฟังก์ชันเพื่อทำการจำลองและทดสอบการประเมินความเสี่ยงรายบุคคลแบบเชื่อมโยงกันทั้งหมด
    (Employee Data -> Prediction -> Local SHAP -> Recommendation -> HR Action View)
    """
    # 1. ดึงข้อมูลดิบและผลการคาดการณ์
    employee_data = X_test_df.loc[[employee_idx]]
    actual_label = "Yes" if y_test_series.loc[employee_idx] == 1 else "No"
    predicted_prob = risk_report_df.loc[employee_idx, 'Predicted_Probability']
    predicted_label = "Yes" if risk_report_df.loc[employee_idx, 'Predicted_Attrition'] == 1 else "No"
    recommendations = risk_report_df.loc[employee_idx, 'Actionable_Recommendations']

    print("="*80)
    print(f"🏥 HR INTERVENTION REPORT: EMPLOYEE CASE STUDY (Index: {employee_idx})")
    print("="*80)
    print(f"[STATUS] Actual Attrition: {actual_label} | Predicted Attrition: {predicted_label} ({predicted_prob*100:.1f}% Probability)")
    print("-"*80)
    print("📌 EMPLOYEE DEMOGRAPHICS & KEY METRICS:")
    print(f"- Age: {employee_data['Age'].values[0]} years")
    print(f"- Monthly Income: ${employee_data['MonthlyIncome'].values[0]:,.2f}")
    print(f"- Distance From Home: {employee_data['DistanceFromHome'].values[0]} miles")
    print(f"- Environment Satisfaction Score: {employee_data['EnvironmentSatisfaction'].values[0]} / 4")
    print(f"- Working Overtime?: {'Yes' if employee_data['OverTime_Yes'].values[0] == 1 else 'No'}")
    print("-"*80)

    # 2. คำนวณ Local SHAP Value
    local_shap = shap_explainer(employee_data)
    # เราสนใจคลาส 1 (Attrition = Yes)
    if len(local_shap.shape) == 3:
        local_shap_values = local_shap.values[0, :, 1]
        base_val = local_shap.base_values[0, 1]
    else:
        local_shap_values = local_shap.values[0, :]
        base_val = local_shap.base_values[0]

    # ดึง Top 3 ฟีเจอร์ที่เพิ่มความเสี่ยงมากที่สุด (ค่าบวกสูง)
    shap_df = pd.DataFrame({
        'Feature': employee_data.columns,
        'SHAP_Value': local_shap_values,
        'Value': employee_data.values[0]
    }).sort_values(by='SHAP_Value', ascending=False)

    print("🔍 LOCAL SHAP EXPLANATION (Top 3 Risk Drivers):")
    top_drivers = shap_df[shap_df['SHAP_Value'] > 0].head(3)
    for idx_row, row in top_drivers.iterrows():
        print(f"  ⚠️ Feature ‘{row['Feature']}’ (Value: {row['Value']}) -> เพิ่มโอกาสลาออก {row['SHAP_Value']*100:+.2f}%")
    print("-"*80)

    print("📋 ACTIONABLE RULE-BASED RECOMMENDATIONS:")
    print(recommendations)
    print("="*80)

# ทดสอบรันระบบกับเคสพนักงานกลุ่มเสี่ยงสูงรหัส Index: 688
verify_individual_risk_path(
    employee_idx=688,
    model=smote_rf,
    X_test_df=X_test,
    y_test_series=y_test,
    risk_report_df=risk_report,
    shap_explainer=explainer
)
```

### 📊 การตรวจสอบคุณภาพข้อมูลและการประเมินอคติ (Data Quality & Bias Audit)
เราจะทำการประเมินชุดข้อมูล IBM HR Analytics ใน 5 มิติสำคัญเพื่อวิเคราะห์ความน่าเชื่อถือและความเหมาะสมในการนำไปสร้างโมเดลทำนายการลาออกเชิงรุก

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# --- 1. Completeness & Consistency Audit ---
total_rows = len(df)
completeness_report = pd.DataFrame({
    'Non-Null Count': df.notnull().sum(),
    'Null Count': df.isnull().sum(),
    'Completeness (%)': (df.notnull().sum() / total_rows) * 100
})

print("=== [1] Completeness Audit (ความสมบูรณ์ของข้อมูล) ===")
display(completeness_report.head(10))

# ตรวจสอบความสอดคล้องเชิงตรรกะ (Consistency)
print("\n=== [2] Consistency Audit (ความสอดคล้องเชิงตรรกะ) ===")
# ตัวอย่าง: ตรวจสอบว่าอายุงานทั้งหมดต้องไม่ต่ำกว่าอายุงานในตำแหน่งปัจจุบัน
consistent_working_years = (df['TotalWorkingYears'] >= df['YearsAtCompany']).all()
consistent_company_years = (df['YearsAtCompany'] >= df['YearsInCurrentRole']).all()

print(f"- อายุงานรวมมากกว่าหรือเท่ากับอายุงานที่บริษัทปัจจุบัน: {consistent_working_years}")
print(f"- อายุงานที่บริษัทปัจจุบันมากกว่าหรือเท่ากับอายุงานในตำแหน่งปัจจุบัน: {consistent_company_years}")
```

```python
# --- 2. Accuracy, Validity & Representativeness ---
print("=== [3] Accuracy / Validity Audit (ความถูกต้องของข้อมูลขอบเขตเชิงตัวเลข) ===")
# ตรวจสอบคอลัมน์ที่เป็นไปไม่ได้ เช่น อายุน้อยกว่า 18 ปี หรือ ชั่วโมงงานมาตรฐานที่แปลกไป
invalid_age = df[df['Age'] < 18]
inconsistent_over18 = df[(df['Age'] >= 18) & (df['Over18'] != 'Y')]

print(f"- จำนวนพนักงานที่มีอายุน้อยกว่า 18 ปี: {len(invalid_age)} คน")
print(f"- จำนวนพนักงานที่อายุมากกว่า 18 แต่ Over18 ไม่ใช่ 'Y': {len(inconsistent_over18)} คน")
print(f"- ตรวจสอบค่าจำเพาะของคอลัมน์ EmployeeCount: {df['EmployeeCount'].unique()}")
print(f"- ตรวจสอบค่าจำเพาะของคอลัมน์ StandardHours: {df['StandardHours'].unique()}")

# --- 3. Representativeness & Bias/Fairness ---
print("\n=== [4] Representativeness & Bias Audit (ความคุ้มครองและความยุติธรรมของข้อมูล) ===")
# ตรวจสอบการกระจายทางประชากรศาสตร์เพื่อประเมินความสมดุลและความเท่าเทียม
gender_dist = df['Gender'].value_counts(normalize=True) * 100
age_mean = df['Age'].mean()

print("สัดส่วนเพศพนักงานในองค์กร:")
display(pd.DataFrame({
    'Count': df['Gender'].value_counts(),
    'Percentage (%)': gender_dist
}))

# วิเคราะห์สัดส่วนการลาออกจำแนกตามกลุ่มประชากรที่ละเอียดอ่อน (Sensitive Attributes เช่น เพศ)
attrition_by_gender = df.groupby('Gender')['Attrition'].value_counts(normalize=True).unstack() * 100
print("\nสัดส่วนการลาออกจำแนกตามเพศ (%):")
display(attrition_by_gender)
```

```python
# แสดงผลกราฟิกเพื่อวิเคราะห์ Representativeness และ Bias
plt.figure(figsize=(14, 5))

# 1. การกระจายของอายุพนักงาน (Representativeness)
plt.subplot(1, 2, 1)
sns.histplot(df['Age'], kde=True, color='skyblue', bins=20)
plt.title('Age Distribution (Representativeness)')
plt.xlabel('Age')
plt.ylabel('Count')

# 2. เปรียบเทียบอัตราการลาออกตามเพศ (Bias/Fairness Evaluation)
plt.subplot(1, 2, 2)
sns.countplot(data=df, x='Gender', hue='Attrition', palette='Set2')
plt.title('Attrition Rate by Gender')
plt.xlabel('Gender')
plt.ylabel('Count')

plt.tight_layout()
plt.show()
```
